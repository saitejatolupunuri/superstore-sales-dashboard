```python
# Filling forward fill() for some essential columns as a part of cleaning.
df['Device_Type'] = df['Device_Type'].ffill()
df[['Subscription_Status', 'Preferred_Content_Type', 'Monthly_Income_Bracket', 'Country','Peak_Usage_Hours']] = df[['Subscription_Status', 'Preferred_Content_Type', 'Monthly_Income_Bracket', 'Country','Peak_Usage_Hours']].ffill()
```

Explaination:
The code uses ffill() (forward fill) to replace NaN values in multiple columns with the most recent valid entry.

Insightful Analysis: 
Forward filling (ffill()) ensures missing values are replaced with the most recent valid entry, maintaining data continuity in categorical fields.'''

```python
## Handling Missing Values in(Age_Group)
def fill_missing_age(row):
    if pd.isnull(row['Age_Group']):
        if row['Age_Group'] == '13-28':
            return "Gen Z"
        elif row['Age_Group'] == '29-44':
            return "Gen Y"
        elif row['Age_Group'] == '45-60':
            return "Gen X"
        elif row['Age_Group'] == '61-100':
            return "Boomer"
        else:
            return "BabyBoomer"  # Default for 61+
    else:
        return row['Age_Group']

df['Age_Group'] = df.apply(fill_missing_age, axis=1)
```

Explaination: 
The function attempts to fill missing values in the Age_Group column by assigning a generational category based on predefined age ranges.
    
Insightful Analysis: 
This function attempts to classify missing values in 'Age_Group' based on predefined age ranges. However, it correctly checks if 'Age_Group' is 'NaN'
then function will perform the analysis and drop them in correct Age_Gropu as per the given Age conditions.'''

```python
#(Outlier Detection) using IQR Method
print("\nOutlier Detection using IQR Method\n")

# Calculate Q1 and Q3
Q1 = df['Data_Consumption_MB'].quantile(0.25)
Q3 = df['Data_Consumption_MB'].quantile(0.75)

# Calculate IQR
IQR = Q3 - Q1

# Define the bounds for outliers
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Identify outliers
outliers_iqr = df[(df['Data_Consumption_MB'] < lower_bound) | (df['Data_Consumption_MB'] > upper_bound)]
print(f"Number of outliers detected using IQR: {outliers_iqr.shape[0]}")

# Scatter plot to visualize outliers
plt.figure(figsize=(10, 5))
sns.scatterplot(x=df.index, y=df['Data_Consumption_MB'], alpha=0.5, label="Normal Data")
sns.scatterplot(x=outliers_iqr.index, y=outliers_iqr['Data_Consumption_MB'], color='red', label="Outliers")
plt.xlabel("Index")
plt.ylabel("Data Consumption (MB)")
plt.title("Outlier Detection - Data Consumption (IQR Method)")
plt.legend()
plt.show()
```

Explaination:
This code identifies outliers in the Data_Consumption_MB column by employing the Interquartile Range (IQR) method:The calculations for Q1 (25th percentile) and Q3 (75th percentile) are performed.
IQR = Q3 - Q1 is calculated to assess the distribution of the data.
The lower and upper limits are specified as Q1 - 1.5 * IQR and Q3 + 1.5 * IQR, respectively. Values falling outside these limits are marked as outliers and recorded. A scatter plot represents typical data points (blue) and anomalies (red).

Insightful Analysis:
The IQR approach identifies extremely high and low data users, assisting in recognizing power users, inactive users, or possible data irregularities. This understanding allows for improved network efficiency, focused marketing, and customized subscription options. Effective management of outliers guarantees precise predictive models and dependable business choices for user involvement and retention.
